```python
img_nx_boat=cv2.imread(imagepath+'New_Zealand_Boat.jpg',cv2.IMREAD_COLOR)
# Image(img_nx_boat)
img_nx_boat=img_nx_boat[:,:,::-1]
plt.imshow(img_nx_boat)
```

```python
#Crop out the middle region of the image
crooped_region=img_nx_boat[200:400,300:600]
plt.imshow(crooped_region)
```

```python
resize_cropped_region=cv2.resize(img_nx_boat,None,fx=1,fy=1)
plt.imshow(resize_cropped_region)
cv2.imwrite((imagepath+'img_nx_boat.png'),resize_cropped_region[:,:,::-1])
```

```python
#Medthod 2: Specifying exact size of the output image
desired_height=100
desired_width=100
dim=(desired_height,desired_width)
resize_cropped=cv2.resize(img_nx_boat,dim,interpolation=cv2.INTER_AREA)
plt.imshow(resize_cropped)
```

```python
#Resize while maintaining aspect ratio
desired_width=100
aspect_ration=desired_width/resize_cropped.shape[1]
desired_height=int(resize_cropped.shape[0]*aspect_ration)
dim=(desired_height,desired_width)
resize_crop=cv2.resize(img_nx_boat,dim,interpolation=cv2.INTER_LINEAR)
plt.imshow(resize_crop)

# Swap channel order
resized_cropped_region_2x = resize_crop[:, :, ::-1]

# Save resized image to disk
cv2.imwrite("resized_cropped_region_2x.png", resize_crop)

# Display the cropped and resized image
Image(filename="resized_cropped_region_2x.png")
```

```python
# Flipping Images
img_nx_boat_hor=cv2.flip(img_nx_boat,1)
img_nx_boat_ver=cv2.flip(img_nx_boat,0)
img_nx_boat_both=cv2.flip(img_nx_boat,-1)

plt.figure(figsize=(18,5))
plt.subplot(141);plt.imshow(img_nx_boat_hor);plt.title("horizontal image")
plt.subplot(142);plt.imshow(img_nx_boat_ver);plt.title("vertical image")
plt.subplot(143);plt.imshow(img_nx_boat_both);plt.title("both flip images")
plt.subplot(144);plt.imshow(img_nx_boat)
```
