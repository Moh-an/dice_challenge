```python
cb_coke=cv2.imread(imagepath+'coca-cola-logo.png',1)

print("Image H, W, C: ",cb_coke.shape)
print("type of image ",cb_coke.dtype)

# display the image by matplot
plt.imshow(cb_coke)
```

```python
# reverse channel to show in correct color
coke_image_reversed_channel=cb_coke[:,:,-1]
plt.figure(figsize=[20,3])

plt.subplot(141)
plt.imshow(coke_image_reversed_channel)
plt.title('green image')
print()

coke_image_reversed_channel_corrected=cb_coke[:,:,::-1]
plt.subplot(142)
plt.imshow(coke_image_reversed_channel_corrected)
plt.title('orginal image')
print()
```

```python
# Split the image into bgr format
cb_img_nz_bgr=cv2.imread(imagepath+'New_Zealand_Lake.jpg',cv2.IMREAD_COLOR)
b,g,r=cv2.split(cb_img_nz_bgr)

# show the channels
plt.figure(figsize=[10,6])
plt.subplot(141);plt.imshow(r,cmap='grey'),plt.title('Red Channel')
plt.subplot(142);plt.imshow(g,cmap='gray'),plt.title('Green Channel')
plt.subplot(143);plt.imshow(b,cmap='gray'),plt.title('Blue Channel')

# merge the channel
nz_merger=cv2.merge((b,g,r))
# plt.subplot(144);plt.imshow(nz_merger);plt.title('Merge channel')
plt.subplot(144)
plt.imshow(nz_merger[:,:,::-1]);
plt.title("correct Merge channel")
```

```python
# OpenCV stores color channels in a differnet order than most other applications (BGR vs RGB).
img_nz_co=cv2.cvtColor(cb_img_nz_bgr,cv2.COLOR_RGB2BGR)
plt.figure(figsize=[10,2])
plt.subplot(131)
plt.imshow(img_nz_co)

plt.subplot(132)
plt.imshow(cv2.cvtColor(img_nz_co,cv2.COLOR_BGR2RGB))

```

```python
img_NZ_rgb = cv2.cvtColor(cb_img_nz_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_NZ_rgb)
```

```python
# Changing to HSV color
img_hsv=cv2.cvtColor(cb_img_nz_bgr,cv2.COLOR_RGB2HSV)
h,s,v=cv2.split(img_hsv)

# show the channel
plt.figure(figsize=[20,5])
plt.subplot(141);plt.imshow(h,cmap='grey');plt.title("H Channel")
plt.subplot(142);plt.imshow(s,cmap='gray');plt.title("S Channel")
plt.subplot(143);plt.imshow(v,cmap='gray');plt.title("V Channel")
plt.subplot(144);plt.imshow(cb_img_nz_bgr);plt.title("Original image")
```

```python
# modifying new channel
h_new= h+10
plt.figure(figsize=[20,5])
plt.subplot(141);plt.imshow(h_new,cmap='grey');plt.title("H Channel")
plt.subplot(142);plt.imshow(s+230,cmap='gray');plt.title("S Channel")
plt.subplot(143);plt.imshow(v+20,cmap='gray');plt.title("V Channel")
plt.subplot(144);plt.imshow(cb_img_nz_bgr);plt.title("Original image")
```
